```python
sales = float(input("Enter sales value ($): "))

cogs = 0.4 * sales
gross_profit = sales - cogs

print(f"Sales: {sales}$")
print(f"Cost of Goods Sold: {cogs}$")
print(f"Gross Profit: {gross_profit}$")
```

```python
sales = float(input("Enter sales value ($): "))
tax_rate = float(input("Enter tax rate (%): "))

cogs = 0.4 * sales
gross_profit = sales - cogs
taxes = gross_profit * tax_rate
net_income = gross_profit - taxes

print(f"Sales: {sales}$")
print(f"Cost of Goods Sold: {cogs}$")
print(f"Gross Profit: {gross_profit}$")
print(f"Taxes: {taxes}$")
print(f"Net Income: {net_income}$")
```

```python
import math
ticketPrice = 10
defaultAudience = 20

print("Ad$\tProfit")
for adMoney in range(0, 201, 25):
  revenue = (defaultAudience + 2 * round(adMoney ** 0.5)) * ticketPrice
  costs = 200 + adMoney
  profit = revenue - costs
  print(f"{float(adMoney)}\t{(profit)}")
```

```python
import random

play = input("Do you want to play a game (y/n)? ")

while play == 'y':
  player = input("Choose b/w rock, paper, scissors: ").lower()
  computer = random.choice(['rock', 'paper', 'scissors'])

  print(f"Player chooses: {player}\nComputer chooses: {computer}")

  if player == computer:
    print("Draw")
  elif (
      (player == "rock" and computer == "scissors") or
      (player == "scissors" and computer == "paper") or
      (player == "paper" and computer == "rock")
      ):
    print("Player wins")
  else:
    print("Computer wins")

play = input("Do you want to continue (y/n)? ")
```

```python
L = []
player = int(input("Enter any number"))
while player > 0:
  print(L)
  L.append(player)
  player = int(input("Enter any number"))
  print(L.sort(reverse=True))
  print(L)
```

```python
numbers = []

num = int(input('Enter a number:'))
while num != 0:
  numbers.append(num)
  num = int(input('Enter another number:'))

print(numbers.sort(reverse=True))
print(numbers)
```
