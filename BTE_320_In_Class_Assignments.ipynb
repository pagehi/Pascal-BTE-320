```python
def getValues(n):
  assert isinstance(n, int), "'n' must be an integer"
  items = []

  for i in range(n):
    items.append(float(input("Enter a value:")))

  if len(items) < 2:
    raise ValueError("the length of list 'items' cannot be < 2")
  return items#items[2:]

def calcAverage(alist):
  assert isinstance(alist, list), "'alist' must be a list"
  try:
    return sum(alist) /len(alist)
  except Exception as e:
    print(f"Error: {e}")


def main():
  n = int(input("Enter a number of items to be added to the list: "))
  alist = getValues(n)
  avg = calcAverage(alist)
  print(f"Average weight: {avg} lbs.")

n = int(input("How many values? "))
alist = getValues(n)

average = calcAverage(alist)

print("Average:", average)
```

```python
class Greeting:
    def __init__(self, first, last):
        self.first = first
        self.last = last

    def display(self, formal):
        if formal:
            print(f"{self.last}, {self.first}")
        else:
            print(self.first)

    def __str__(self):
        return "Instance of class Greeting"

name = Greeting("John", "Doe")
print(name)
name.display(True)
name.display(False)
```

```python
class Student:
  def __init__(self, name, number):
    self.__name = name
    self.number = number
    self.__courses = []

  def get_name(self):
    return self.__name

  def get_courses(self):
    return self.__courses

  def set_name(self, newName):
    self.__name = newName

  def __len__(self):
    return len(self.__courses)

  def enroll(self, newCourse):
    if newCourse not in self.__courses:
      self.__courses.append(newCourse)
    else:
      print(f"Student {self.__name} is already enrolled in {newCourse}")
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance(self, other):
        dx = (self.x - other.x) ** 2
        dy = (self.y - other.y) ** 2
        return (dx + dy) ** 0.5

class Rectangle:
    def __init__(self):
        self.p1 = Point(0, 0)
        self.p2 = Point(3, 0)
        self.p3 = Point(0, 3)
        self.p4 = Point(3, 3)
        self.width = self.p1.distance(self.p2)
        self.height = self.p1.distance(self.p3)

    def area(self):
        return self.width * self.height

rect = Rectangle()
print("Width:", rect.width)
print("Height:", rect.height)
print("Area:", rect.area())
```

```python
import math
class Employee:
  def __init__(self, name, wage, hours):
    self.name = name
    self.hourly_wage = wage
    self.hours = hours

  def get_wage(self):
    return self.hourly_wage

  def set_wage(self, newWage):
    if newWage <= 0:
      print("Negative wage values not allowed.")

    else:
      self.__hourly_wage = newWage

  def calculate_pay(self):
    return self.hourly_wage * self.hours

  def display_info(self):
    print(f"Employee: {self.name}")
    print(f"Hourly wages: {self.hourly_wage}$/hour")

class Manager(Employee):
  def __init__(self, name, wage, hours, bonus):
    super().__init__(name, wage, hours)
    self.bonus = bonus

  def calculate_pay(self):
    return (1+ self.bonus) * self.get_wage() * self.hours

employee = Employee("Bob", 50, 10)
manager = Manager("Alice", 100, 15, 0.2)
employee.display_info()
manager.display_info()
print()
print(f"Employee {employee.name} weekly salary is: {employee.calculate_pay()}$")
print(f"Manager {manager.name} weekly salary is: {manager.calculate_pay()}$ that includes a {manager.bonus * 100}% bonus")
```
